```python
import glob, os, shutil
R = "results/outputs"
def copy(pattern, dest, limit=None):
    os.makedirs(dest, exist_ok=True)
    for f in sorted(glob.glob(pattern))[:limit]:
        shutil.copy(f, dest)

copy(f"{OUT}/Tuning/*.csv", f"{R}/tuning")
copy(f"{OUT}/Tuning/*.json", f"{R}/tuning")
copy(f"{OUT}/Tuning/*.png", f"{R}/tuning")
copy(f"{OUT}/WD-01-evaluation/evaluation_metrics.json", f"{R}/evaluation/WD-01")
copy(f"{OUT}/mlflow/runs_summary.csv", f"{R}/tracking")
copy(f"{OUT}/mlflow/tracked_runs_curves.png", f"{R}/tracking")
copy(f"{OUT}/GradCAM/Fine-Tuning/gradcam_summary.csv", f"{R}/gradcam/fine_tuning")
copy(f"{OUT}/GradCAM/Fine-Tuning/correct/*.png", f"{R}/gradcam/fine_tuning/correct", 4)
copy(f"{OUT}/GradCAM/Fine-Tuning/incorrect/*.png", f"{R}/gradcam/fine_tuning/incorrect", 8)

for root, _, files in os.walk("results"):
    for f in files: print(os.path.join(root, f))
shutil.make_archive("workstream_d_results", "zip", "results")
from google.colab import files
files.download("workstream_d_results.zip")
```
